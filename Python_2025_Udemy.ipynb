### Funkcijų kūrimas ir naudojimas Python kalboje

Funkcija Python kalboje yra pavadintas kodo blokas, kuris atlieka konkrečią užduotį. Funkcijos padeda organizuoti kodą, jį pakartotinai naudoti ir padaryti programą lengviau skaitomą.

**Pagrindinė sintaksė:**
```python
def funkcijos_pavadinimas(parametras1, parametras2):
    # Kodo blokas, kurį atlieka funkcija
    # ...
    return rezultatas # Neprivaloma: grąžina reikšmę
```

**Paaiškinimas:**
*   `def`: Tai raktinis žodis, kuris nurodo, kad apibrėžiame funkciją.
*   `funkcijos_pavadinimas`: Jūsų pasirinktas unikalus funkcijos pavadinimas. Rekomenduojama naudoti mažąsias raides ir atskirti žodžius pabraukimu (`_`).
*   `(parametras1, parametras2)`: Tai yra neprivalomi parametrai (argumentai), kuriuos funkcija gali priimti. Jie nurodomi skliausteliuose. Funkcijos viduje šie parametrai veikia kaip vietiniai kintamieji. Jei funkcija nepriima jokių argumentų, skliausteliai lieka tušti (`()`).
*   `:`: Dvitaškis žymi funkcijos antraštės pabaigą ir kodo bloko pradžią.
*   **Įtraukos:** Visi kodo bloko sakiniai turi būti vienodai įtraukti (dažniausiai keturiais tarpais), kad Python žinotų, jog jie priklauso tai funkcijai.
*   `return rezultatas`: `return` sakinys yra neprivalomas. Jei jis naudojamas, funkcija grąžina nurodytą reikšmę. Jei `return` nėra, funkcija automatiškai grąžina `None`.

**Funkcijos iškvietimas (naudojimas):**
Norėdami paleisti funkcijos kodą, tiesiog parašykite jos pavadinimą, po kurio seka skliausteliai su reikalingais argumentais (jei tokių yra):

```python
funkcijos_pavadinimas(reiksme1, reiksme2)
```

### Enumerate Function

The `enumerate()` function adds a counter to an iterable and returns it as an enumerate object. This object can then be used directly in loops to get both the index and the item at each iteration. It's particularly useful when you need to know the position of an element while processing it.

```python
word = 'abcde'
for index, letter in enumerate(word):
  print(index, letter)
```

### Zip Function

The `zip()` function is considered conceptually 'opposite' to `enumerate()` because:

*   **`enumerate()`:** Takes *one* iterable and *adds an index* to it, producing pairs of `(index, item)`.
*   **`zip()`:** Takes *multiple* iterables and *combines them element-wise*, creating tuples where each tuple contains elements from the corresponding positions in the input iterables. It stops when the shortest iterable is exhausted.

`zip()` allows you to iterate over several iterables at once, pairing up their elements.

````
list_a = [1, 2, 3]
list_b = ['a', 'b', 'c']
list_c = [True, False, True]

# Using zip to combine elements from multiple lists
for item_a, item_b, item_c in zip(list_a, list_b, list_c):
  print(f"Combined: {item_a}, {item_b}, {item_c}")

# You can also cast zip to a list of tuples
zipped_lists = list(zip(list_a, list_b))
print(f"\nZipped lists (list_a, list_b): {zipped_lists}")
````
